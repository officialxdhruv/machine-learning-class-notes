```python
# ===============================
# ANN Implementation (Single File)
# ===============================

# Step 1: Import Libraries
import numpy as np
import matplotlib.pyplot as plt

from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import Adam

# -------------------------------
# Step 2: Dataset (XOR Problem)
# -------------------------------
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([0, 1, 1, 0])

# -------------------------------
# Step 3: Build ANN Model
# -------------------------------
model = Sequential()

# Hidden Layer
model.add(Dense(units=4, activation='relu', input_dim=2))

# Output Layer
model.add(Dense(units=1, activation='sigmoid'))

# -------------------------------
# Step 4: Compile Model
# -------------------------------
model.compile(
    optimizer=Adam(learning_rate=0.01),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# -------------------------------
# Step 5: Train Model
# -------------------------------
history = model.fit(
    X, y,
    epochs=200,
    verbose=0   # change to 1 if you want to see training logs
)

# -------------------------------
# Step 6: Evaluate Model
# -------------------------------
loss, accuracy = model.evaluate(X, y, verbose=0)
print(f"Accuracy: {accuracy:.4f}")

# -------------------------------
# Step 7: Predictions
# -------------------------------
predictions = model.predict(X)
pred_labels = (predictions > 0.5).astype(int)

print("\nPredictions (Probabilities):")
print(predictions)

print("\nPredicted Labels:")
print(pred_labels)

print("\nActual Labels:")
print(y)

# -------------------------------
# Step 8: Visualization
# -------------------------------
plt.plot(history.history['loss'])
plt.title('Loss vs Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.grid()
plt.show()

# -------------------------------
# Step 9: Model Summary
# -------------------------------
print("\nModel Summary:")
model.summary()
```
