```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from keras.models import Sequential
from keras.layers import Dense
from keras.utils import to_categorical
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../datasets/Iris.csv")
df.head()
```

```python
df.info()
```

```python
df = df.drop('Id', axis='columns')
```

```python
x = df.drop(columns=["Sp","Species"], axis='columns')
x.head()
```

```python
y = df.Sp
y.head()
```

```python
y_encoded = to_categorical(y)
y_encoded
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y_encoded, test_size=0.3, random_state=42
)
```

```python
# Model
model = Sequential([
    Dense(10, activation='relu', input_shape=(4,)), #hhidden layer
    # Dense(10, activation='relu'),
    # Dense(5, activation='relu'),
    Dense(3, activation='softmax')  # 3 output classes
])

model.compile(optimizer='adam', loss='mse', metrics=['accuracy'])

model.fit(x_train, y_train, epochs=30, verbose=1)
```

```python
# Evaluate
loss, acc = model.evaluate(x_test, y_test)
print(f"Accuracy: {acc:.2f}")
```

```python
# Predict
y_pred_probs = model.predict(x_test)
y_pred = np.argmax(y_pred_probs, axis=1)
y_true = np.argmax(y_test, axis=1)
```

```python
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, cmap='Blues', fmt='d', xticklabels=['0','1','2'], yticklabels=['0','1','2'])
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()
```

```python
# Step 3: Classification Report
print("\nClassification Report:\n")
print(classification_report(y_true, y_pred, target_names=['0','1','2']))
```
