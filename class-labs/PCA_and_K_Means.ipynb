```python
import numpy as np
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Small dataset
X = np.array([[2,4],
              [4,6],
              [6,8],
              [8,10]])

# Apply PCA (reduce to 1 dimension)
pca = PCA(n_components=1)
X_pca = pca.fit_transform(X)

print("Original Data:\n", X)
print("\nReduced Data (1D using PCA):\n", X_pca)

# Visualization
plt.scatter(X[:,0], X[:,1])
plt.title("Original Data")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()

plt.scatter(X_pca, np.zeros(len(X_pca)))
plt.title("Data after PCA (1D)")
plt.show()
```

```python
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Small dataset
X = np.array([[1,1],
              [2,1],
              [4,3],
              [5,4]])

# Apply K-Means
kmeans = KMeans(n_clusters=2, random_state=0)
kmeans.fit(X)

labels = kmeans.labels_
centroids = kmeans.cluster_centers_

print("Cluster Labels:", labels)
print("Centroids:\n", centroids)

# Visualization
plt.scatter(X[:,0], X[:,1], c=labels, cmap='rainbow')
plt.scatter(centroids[:,0], centroids[:,1],
            marker='x', s=200)
plt.title("K-Means Clustering")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

# Small dataset
X = np.array([[2,4],
              [4,6],
              [6,8],
              [8,10]])

# Step-1: Apply PCA (2D → 1D)
pca = PCA(n_components=1)
X_reduced = pca.fit_transform(X)

# Step-2: Apply K-Means
kmeans = KMeans(n_clusters=2, random_state=0)
labels = kmeans.fit_predict(X_reduced)
centroids = kmeans.cluster_centers_

# Step-3: Plotting
plt.figure()

# Plot cluster-1 points
plt.scatter(X_reduced[labels==0],
            np.zeros_like(X_reduced[labels==0]),
            marker='o',
            label='Cluster 1')

# Plot cluster-2 points
plt.scatter(X_reduced[labels==1],
            np.zeros_like(X_reduced[labels==1]),
            marker='s',
            label='Cluster 2')

# Plot centroids
plt.scatter(centroids,
            np.zeros_like(centroids),
            marker='x',
            s=200,
            label='Centroids')

plt.title("PCA + K-Means Clustering")
plt.xlabel("Principal Component-1")
plt.yticks([])

plt.legend()
plt.grid()
plt.show()
```
