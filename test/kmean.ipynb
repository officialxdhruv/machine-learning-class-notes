```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Data
X = np.array([
    [1,2], [2,1], [2,3], [3,2],
    [8,8], [9,8], [8,9], [9,9],
    [25,30], [24,29]
])

# KMeans model
kmeans = KMeans(
    n_clusters=3,
    init=np.array([
        [1,2],
        [8,8],
        [25,30]
    ]),
    n_init=1
)

# Fit
kmeans.fit(X)

labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# Plot points
plt.scatter(X[:, 0], X[:, 1], c=labels)

# Plot centroids
plt.scatter(centroids[:, 0], centroids[:, 1], marker='x', s=200)

# Add labels to points (optional but good for exams)
# for i, (x, y) in enumerate(X):
#     plt.text(x - 0.5, y + 0.2, f"P{i+1}")

# Title and axis
plt.title("K-Means Clustering")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

# Show grid
plt.grid()

# Show plot
plt.show()
```
